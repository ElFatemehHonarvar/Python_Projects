**user guess**
- ```1 <user guess <100```
- user guess must be an int (not char or any other objects)
- "q" means the game should be quit

**game**
- give hint to the user to help her is the guess is lower or higher than actual number
- give user a score and reduce it when the her guess is wrong


````python

import random

start = 1
end = 100
def validate_guess(user_guess, start, end):

  if not user_guess.isdigit():
    print(" Invalid input, your input is not integer, Please enter a number between 1 to 100")
    return False

  user_guess = int(user_guess)
  if (int(user_guess) < start) or (int(user_guess) > end):
    print("Invalid input, Please enter a number between 1 to 100")
    return False

  return True



def game(start, end):
  number = random.randint(start, end)
  score = 10


  while True :
    user_guess = input("guess a number : ")

    if user_guess == "q":
      print("You quit the game")
      break


    if validate_guess(user_guess, start, end):

      user_guess = int(user_guess)
      if user_guess == number:
        print("Horaaaaa you win")
        print(f"your score is : {score}")

        play_again = input("Would you like to play again? select y or n ")
        if play_again == "y":

          number = random.randint(1, 100)
          score = 10
          continue

        break

      if score < 8:
        print("sorry you lose the game")
        
        play_again = input("Would you like to play again? select y or n ")
        if play_again == "y":

          number = random.randint(1, 100)
          score = 10
          continue
        break

      if user_guess > number:
        score -= 1
        print(f"you should select a number which is smaller than {user_guess}")

      else:
        score -= 1
        print(f"you should select a number which is bigger than {user_guess}")

    else:
      continue



game(start, end)
# if __name__ == "__main__":
#   game()


````

````python
def validate_guess():
  while True:
    user_guess = input("guess a number : ")
    if user_guess == "q":
      break

    elif not user_guess.isdigit():
      print(" Invalid input, your input is not integer, Please enter a number between 1 to 100")
      continue

    elif (int(user_guess) < 1) or (int(user_guess) > 100):
      print("Invalid input, Please enter a number between 1 to 100")
      continue

    print(f" your guess is {user_guess}")
    return int(user_guess)


````
