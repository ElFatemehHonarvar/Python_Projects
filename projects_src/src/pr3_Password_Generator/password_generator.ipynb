Create a Python application that generates three types of passwords:

Random Passwords
Memorable Passwords
Pin Codes

Requirements
Your application should have the following features:

A RandomPasswordGenerator that generates a completely *random password* of a *specified length*, optionally including *numbers and symbols*.
A MemorablePasswordGenerator that creates a password made up of a *set number of randomly chosen words from the NLTK* English language corpus. It should also *optionally separate the words with a chosen separator* and *make the words capitalized*.
A PinCodeGenerator that creates a *numeric password of a specified length*.
Each generator class should inherit from a base PasswordGenerator class and override the base class's generate() function to provide their unique password generation functionality.

- RandomPasswordGenerator
    - nums & symbols (optional)
    - specified length

- MemorablePasswordGenerator
    - words from NLTK
    - set number of randomly
    - word separator (optional)
    - captalization (optional)

- PinCodeGenerator
    - nums
    - specified length

parent class :
PasswordGenerator -> Abstract class

    - generate() function


```python
from abc import ABC , abstractmethod

import random
import string

class PasswordGenerator(ABC):

    @abstractmethod
    def generate(self):
        pass
```

```python
class PinCodeGenerator(PasswordGenerator):

    def __init__(self, length):
        self.length = length

    def generate(self):
            pass_num = "".join(random.choices(string.digits, k=self.length))
            print(pass_num)
```

```python
pin = PinCodeGenerator(5)

pin.generate()
```

```python
class RandomPasswordGenerator(PasswordGenerator):

    def __init__(self):
        self.length = int(input("What is the length of the password"))
        self.include_numbers = input("would you like to include numbers? select y/n")
        self.include_symbols = input("would you like to include symbols? select y/n")


    def generate(self):

        if (self.include_numbers.lower() == "y") and (self.include_symbols.lower() == "y"):
            password = "".join(random.choices(string.ascii_letters + string.digits, k=self.length))
            return password

        elif (self.include_numbers.lower() == "y"):
            password = "".join(random.choices(string.digits, k=self.length))
            return password

        elif (self.include_symbols.lower() == "y"):
            password = "".join(random.choices(string.ascii_letters, k=self.length))
            return password

        else:
            return ValueError()


random_pass = RandomPasswordGenerator()
random_pass.generate()
```

```python

class RandomPasswordGenerator(PasswordGenerator):

    def __init__(self, length, include_numbers=True, include_symbols=True):

        self.length = length
        self.include_numbers = include_numbers
        self.include_symbols = include_symbols
        self.characters = string.ascii_letters

        if self.include_numbers:
            self.characters += string.digits

        if self.include_symbols:
            self.characters += string.punctuation



    def generate(self):

        password = "".join(random.choices(self.characters, k=self.length))
        return password
```

```python
random_pass = RandomPasswordGenerator(10, include_numbers=True, include_symbols=True)
random_pass.generate()
```

```python
import nltk

class MemorablePasswordGenerator(PasswordGenerator):
    def __init__(self):

        self.num_of_words = int(input("set the number of words"))
        self.captalize = input("would you like to captilize words? select y/n")
        self.separator = input(f"select your separator from options : {string.punctuation} unless the default would be set")
        self.vocabulary= nltk.corpus.words.words()

        if self.separator not in string.punctuation:
            self.separator = "-"  # Enter is an issue


    def generate(self):

        words = random.choices(nltk.corpus.words.words(), k=self.num_of_words)
        if (self.captalize == "y"):
            for i in range(len(words)):
                words[i] = words[i].upper()

        return self.separator.join(words)


mem_password = MemorablePasswordGenerator()
mem_password.generate()
```

```python


class MemorablePasswordGenerator(PasswordGenerator):
    def __init__(self, num_of_words, random_captalize, separator, vocabulary=None):

        if vocabulary is None:
            vocabulary = nltk.corpus.words.words()

        self.num_of_words = num_of_words
        self.random_captalize = random_captalize
        self.separator = separator
        self.vocabulary = vocabulary

        if self.separator not in string.punctuation:
            self.separator = "-"  # Enter is an issue

    def generate(self):

        words = random.choices(self.vocabulary, k=self.num_of_words)
        if self.random_captalize:
            upper_words = [word.upper() if random.choice([True, False]) else word.lower() for word in words]

        return self.separator.join(upper_words)
```

```python
mem_password = MemorablePasswordGenerator(5, True, "_" )
mem_password.generate()
```
