Create a Python application that generates three types of passwords:

Random Passwords
Memorable Passwords
Pin Codes

Requirements
Your application should have the following features:

A RandomPasswordGenerator that generates a completely *random password* of a *specified length*, optionally including *numbers and symbols*.
A MemorablePasswordGenerator that creates a password made up of a *set number of randomly chosen words from the NLTK* English language corpus. It should also *optionally separate the words with a chosen separator* and *make the words capitalized*.
A PinCodeGenerator that creates a *numeric password of a specified length*.
Each generator class should inherit from a base PasswordGenerator class and override the base class's generate() function to provide their unique password generation functionality.

- RandomPasswordGenerator
    - nums & symbols (optional)
    - specified length

- MemorablePasswordGenerator
    - words from NLTK
    - set number of randomly
    - word separator (optional)
    - captalization (optional)

- PinCodeGenerator
    - nums
    - specified length

parent class :
PasswordGenerator class
    - generate() function


```python
import random
import string

class PasswordGenerator:

    def __init__(self, length):
        self.length = length

    def generate(self):
        pass
```

```python
class PinCodeGenerator(PasswordGenerator):

    def __init__(self, length):
        super(PinCodeGenerator, self).__init__(length)

    def generate(self):
            pass_num = "".join(random.choices(string.digits, k=self.length))
            print(pass_num)
```

```python
pin = PinCodeGenerator(5)

pin.generate()
```

```python
class RandomPasswordGenerator(PasswordGenerator):

    def __init__(self, length):
        super(RandomPasswordGenerator, self).__init__(length)
        self.numbers = input("would you like to include numbers? select y/n")
        self.symbols = input("would you like to include symbols? select y/n")


    def generate(self):
        if (self.numbers.lower() == "y") and not (self.symbols.lower() == "y"):
            password = "".join(random.choices(string.digits, k=self.length))
            return password

        elif (self.numbers.lower() == "y") and (self.symbols.lower() == "y"):
            password = "".join(random.choices(string.ascii_letters + string.digits, k=self.length))
            return password

        elif not (self.numbers.lower() == "y") and (self.symbols.lower() == "y"):
            password = "".join(random.choices(string.ascii_letters, k=self.length))
            return password

        else:
            return ValueError()


```

```python
random_pass = RandomPasswordGenerator(5)
random_pass.generate()
```

```python
import nltk

class MemorablePasswordGenerator(PasswordGenerator):
    def __init__(self, length):
        super(MemorablePasswordGenerator, self).__init__(length)

        self.captalize = input("would you like to captilize words? select y/n")
        self.separator = input("select your separator from options : (, . - / _) unless enter")

    def generate(self):

        words = random.choices(nltk.corpus.abc.words(), k=self.length)
        print(words)
        if (self.captalize == "y"):
            for i in range(len(words)):
                words[i] = words[i].upper()


        if self.separator in [",", ".", "_", "-", "/"]:
            print(self.separator.join(words))
        
        else:
            print("".join(words)) 
        
```

```python
mem_password = MemorablePasswordGenerator(5)
mem_password.generate()
```
